# params

> Simulation parameters (Table 1 defaults) for explRL agents and training runs.

```python
#| export
from dataclasses import dataclass
```

```python
#| export
@dataclass
class Params:
    nx: int = 96            # basis columns (arena positions) around 360 deg
    delT: float = 0.001     # fixation sample length, s (delta t)
    tS: float = 0.32        # saccade duration, s (Table 1)
    sigF: float = 1.0       # spread of drift diffusion (eta_F)
    aF: float = 0.79        # fixation threshold
    muS: float = 3.89       # lognormal saccade size, deg (phi_S)
    sigmaS: float = 0.54    # lognormal saccade size spread
    Ttot: float = 240.0     # run length, s
    nb: int = 16            # number of basis functions
    kappa: float = 8.0      # basis concentration
    wParams: tuple = (1.0, 10.0, -0.01, 1.0, 0.98, -0.01)  # kF, fMF, f0F, kS, fMS, f0S
    omegaF0: float = 0.1    # initial fixation weights
    omegaS0: float = 0.0    # initial saccade weights
    lr: float = 1.0         # extra learning-rate factor (paper: 1)

    @property
    def dtS(self) -> int:
        "Saccade length in samples."
        return round(self.tS / self.delT)

    def __post_init__(self):
        if self.nx <= 0 or self.nb <= 0:
            raise ValueError('nx and nb must be positive')
        if self.delT <= 0 or self.tS <= 0 or self.sigF <= 0 or self.aF <= 0 or self.Ttot <= 0:
            raise ValueError('delT, tS, sigF, aF, and Ttot must be positive')
        if len(self.wParams) != 6:
            raise ValueError('wParams must have length 6')
```

```python
from fastcore.test import test_eq, test_fail

p = Params()
test_eq(p.nx, 96)
test_eq(p.delT, 0.001)
test_eq(p.tS, 0.32)
test_eq(p.sigF, 1.0)
test_eq(p.aF, 0.79)
test_eq(p.muS, 3.89)
test_eq(p.sigmaS, 0.54)
test_eq(p.Ttot, 240.0)
test_eq(p.nb, 16)
test_eq(p.kappa, 8.0)
test_eq(p.wParams, (1.0, 10.0, -0.01, 1.0, 0.98, -0.01))
test_eq(p.omegaF0, 0.1)
test_eq(p.omegaS0, 0.0)
test_eq(p.lr, 1.0)
```

```python
test_eq(Params().dtS, 320)
```

```python
p = Params(nx=48, lr=0.25)
test_eq(p.nx, 48)
test_eq(p.lr, 0.25)
test_eq(p.dtS, 320)
```

```python
test_fail(lambda: Params(nx=0), contains='positive')
test_fail(lambda: Params(wParams=(1.0, 2.0, 3.0)), contains='wParams')
```
