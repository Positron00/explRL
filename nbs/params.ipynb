```python
#| export
from dataclasses import dataclass
```

```python
#| export
@dataclass
class Params:
    nx: int = 96            # basis columns (arena positions) around 360 deg
    delT: float = 0.001     # fixation sample length, s (delta t)
    tS: float = 0.32        # saccade duration, s (Table 1)
    sigF: float = 1.0       # spread of drift diffusion (eta_F)
    aF: float = 0.79        # fixation threshold
    muS: float = 3.89       # lognormal saccade size, deg (phi_S)
    sigmaS: float = 0.54    # lognormal saccade size spread
    Ttot: float = 240.0     # run length, s
    nb: int = 16            # number of basis functions
    kappa: float = 8.0      # basis concentration
    wParams: tuple = (1.0, 10.0, -0.01, 1.0, 0.98, -0.01)  # kF, fMF, f0F, kS, fMS, f0S
    omegaF0: float = 0.1    # initial fixation weights
    omegaS0: float = 0.0    # initial saccade weights
    lr: float = 1.0         # extra learning-rate factor (paper: 1)
    GS: float = 0.9         # saccade heading gain (Table 1)
    BS: float = 0.0         # saccade baseline direction
    GF: float = 0.8         # fixation heading gain
    BF: float = 0.05        # fixation baseline drift rate
    GJ: float = 0.7         # bump-jump heading gain
    BJ: float = 0.05        # bump-jump baseline probability
    thetaJ: float = 3.141592653589793  # jump size, radians (180 deg)
    lrG: float = 1.0        # goal-heading learning-rate factor
    nCirc: int = 32         # circuit heading grid (Table 1 N)
    kappaEB: float = 3.141592653589793  # EB bump concentration (Table 1 kappa)
    alphaC: float = 0.01    # compass weight learning rate
    alphaG: float = 0.001   # goal weight learning rate
    nuMax: float = 1.1      # fixation output ceiling (Eq. 32)
    dtG: float = 0.1        # goal update step during fixation, s

    @property
    def dtS(self) -> int:
        "Saccade length in samples."
        return round(self.tS / self.delT)

    def __post_init__(self):
        if self.nx <= 0 or self.nb <= 0:
            raise ValueError('nx and nb must be positive')
        if self.delT <= 0 or self.tS <= 0 or self.sigF <= 0 or self.aF <= 0 or self.Ttot <= 0:
            raise ValueError('delT, tS, sigF, aF, and Ttot must be positive')
        if len(self.wParams) != 6:
            raise ValueError('wParams must have length 6')
        if abs(self.GS) / 2 + abs(self.BS) > 0.5:
            raise ValueError('turnProb must stay in [0, 1]')
        if self.BF <= 0 or self.GF < 0:
            raise ValueError('BF must be positive and GF non-negative')
        if self.GJ < 0 or self.BJ < 0 or self.GJ + self.BJ > 1:
            raise ValueError('jumpProb must stay in [0, 1]')
        if self.lrG < 0:
            raise ValueError('lrG must be non-negative')
        if self.nCirc < 4 or self.nCirc % 2:
            raise ValueError('nCirc must be even and at least 4')
        if self.kappaEB <= 0 or self.nuMax <= 0 or self.dtG <= 0:
            raise ValueError('kappaEB, nuMax, and dtG must be positive')
        if self.alphaC < 0 or self.alphaG < 0:
            raise ValueError('alphaC and alphaG must be non-negative')
```

```python
from fastcore.test import test_eq, test_fail, test_close

p = Params()
test_eq(p.nx, 96)
test_eq(p.delT, 0.001)
test_eq(p.tS, 0.32)
test_eq(p.sigF, 1.0)
test_eq(p.aF, 0.79)
test_eq(p.muS, 3.89)
test_eq(p.sigmaS, 0.54)
test_eq(p.Ttot, 240.0)
test_eq(p.nb, 16)
test_eq(p.kappa, 8.0)
test_eq(p.wParams, (1.0, 10.0, -0.01, 1.0, 0.98, -0.01))
test_eq(p.omegaF0, 0.1)
test_eq(p.omegaS0, 0.0)
test_eq(p.lr, 1.0)
test_eq(p.GS, 0.9)
test_eq(p.BS, 0.0)
test_eq(p.GF, 0.8)
test_eq(p.BF, 0.05)
test_eq(p.GJ, 0.7)
test_eq(p.BJ, 0.05)
test_close(p.thetaJ, 3.141592653589793)
test_eq(p.lrG, 1.0)
test_eq(p.nCirc, 32)
test_close(p.kappaEB, 3.141592653589793)
test_eq(p.alphaC, 0.01)
test_eq(p.alphaG, 0.001)
test_eq(p.nuMax, 1.1)
test_eq(p.dtG, 0.1)
```

```python
test_eq(Params().dtS, 320)
```

```python
p = Params(nx=48, lr=0.25)
test_eq(p.nx, 48)
test_eq(p.lr, 0.25)
test_eq(p.dtS, 320)
```

```python
test_fail(lambda: Params(nx=0), contains='positive')
test_fail(lambda: Params(wParams=(1.0, 2.0, 3.0)), contains='wParams')
test_fail(lambda: Params(GS=1.0, BS=0.1), contains='turnProb')
test_fail(lambda: Params(BF=0.0), contains='BF')
test_fail(lambda: Params(GJ=0.6, BJ=0.5), contains='jumpProb')
test_fail(lambda: Params(lrG=-0.1), contains='lrG')
test_fail(lambda: Params(nCirc=3), contains='nCirc')
test_fail(lambda: Params(kappaEB=0.0), contains='kappaEB')
test_fail(lambda: Params(alphaC=-0.01), contains='alphaC')
```
