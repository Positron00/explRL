```python
#| export
def vonMisesBasis(nb: int, nx: int, kappa: float) -> np.ndarray:
    "Returns an (nb, nx) matrix; row i is a von Mises bump centred at 2*pi*i/nb, peak value 1."
    if nb <= 0 or nx <= 0:
        raise ValueError('nb and nx must be positive')
    if kappa <= 0:
        raise ValueError('kappa must be positive')
    i = np.arange(nb)[:, None]
    j = np.arange(nx)[None, :]
    mu_i = 2 * np.pi * i / nb
    theta_j = 2 * np.pi * j / nx
    return np.exp(kappa * (np.cos(theta_j - mu_i) - 1))
```

```python
from fastcore.test import test_eq, test_close, test_fail

nb, nx, kappa = 16, 96, 8.0
g = vonMisesBasis(nb, nx, kappa)
test_eq(g.shape, (nb, nx))
```

```python
test_eq(g.max(), 1.0)
```

```python
for i in range(nb):
    test_eq(int(g[i].argmax()), round(i * nx / nb))
```

```python
shift = nx // nb
for i in range(nb):
    test_close(np.roll(g[i], shift), g[(i + 1) % nb])
```

```python
test_fail(lambda: vonMisesBasis(0, nx, kappa), contains='positive')
test_fail(lambda: vonMisesBasis(nb, 0, kappa), contains='positive')
test_fail(lambda: vonMisesBasis(nb, nx, 0.0), contains='kappa')
```
