```python
#| export
from explRL.bridge_grid import (ALLOW20, ARMS20, C1_CELLS, C1_CLEAN, PARTB_MODELS, PER_CALL4, PILOT20, REUSED20, SEEDS5, SEEDS8,
                                arm20, c1Estimate, c1Grid, c1Range)
from explRL.colony_grid import PILOT_SEED, WORLD18

_L = 'honestPooled, N 64, independent liars'
C1_PILOT_PLAN = dict(
    version=1, frozen=False, plan='specs/multiagent/phase20/', model='claude-haiku-5-5', world=WORLD18, turns=40,
    channel='blackboard', memory='full', effort='low', arms=ARMS20, cells=dict(
        liars={a: s for a, s in C1_CELLS.items()}, clean={a: s for a, s in C1_CLEAN.items()}, pilot=list(PILOT20),
        reused={a: dict(study=st, seeds=SEEDS8) for a, st in REUSED20}),
    seeds=dict(eight=SEEDS8, five=SEEDS5, pilot=PILOT_SEED),
    faults=dict(independent='scripted ClueLiar in the first 16 slots, guards the next 16 (colonyCell)',
                clean='no liars: 64 Claude agents, the 16 guards in the first 16 slots (P20-14)'),
    hypotheses=dict(
        H20_1=f'Gate C1 (i), primary: bridge1 - colonyGuards, {_L}, 8 boards paired on seed: one-sided exact sign-flip > 0, '
              'alpha 0.05 (smallest p 1/256)',
        H20_2='Gate C1 (ii), a margin, not a test: mean(bridge1 - colonyGuards) >= 0.5 x mean(squads - colonyGuards), the same 8 '
              'boards; bootstrap interval; uncorrected sign-flip p of the per-board margin, reported',
        H20_3='win-back (bridge1 - coupled) / (squads - coupled), means over 1800-1804, > 0.5; per-board sign-flip p of '
              '(bridge1 - coupled) - 0.5 (squads - coupled)',
        H20_4='(bridge2 - bridge1) < 0.25 x (squads - coupled), seeds 1800-1804; interval',
        H20_5='false-relay share, bridge1 - bridgeGuard > 0, seeds 1800-1804; also honestPooled, bridgeGuard - bridge1',
        H20_6='part b (descriptive): the gain over the warm start in squads of 4 minus on one board, 32 agents, clean; sign-flip '
              'over the 8 runs (smallest p 1/256); Phase 17 gains +0.078 (4 agents) and +0.006 (32) beside it',
        H20_7='cross-squad contamination per arm, descriptive (exploratory)',
        H20_8='cost without liars (P20-14), descriptive, not in Gate C1: honestPooled and welfareHonest per arm on clean boards; '
              "coupled - squads and bridge1 - squads, colony_report.pairs(rows, arm, m, fault='clean', base='squads'), "
              '1800-1804, bootstrap interval, uncorrected one-sided sign-flip p < 0 (smallest 1/32)'),
    gate=dict(parts=['H20_1', 'H20_2'], alpha=0.05, metric='honestPooled', lead=0.171, bar=0.086,
              passes='squads with bridges become the training world of A1 and C3 (Phase 19, P19-1)'),
    robustness='H20-1 and H20-2 on welfareHonest; an estimate of the opposite sign marks the part not robust',
    inertBridge='if Gate C1 passes and mean(bridge1 - coupled) <= 0 on seeds 1800-1804: stop and ask before A1 adopts bridges',
    families='no Holm: each secondary has 5 boards (smallest p 1/32), reported uncorrected and labelled exploratory',
    coreMeasures='FR-D5: contestColumns per board; gullibility, suppression, verification, collaboration and error correlation '
                 'pooled per arm (rl19_measures.ratio, lags); welfare per agent-turn and per 1,000 tokens; tokens per solved task; '
                 "Gorelick's index per board (rl19_spec.dolIndex); cross-squad contamination; no role persistence (single boards)",
    pilot=dict(cells=list(PILOT20), seed=PILOT_SEED, analysed=False, maxInvalid=0.05, minMirrorPosts=1, minRelays=1,
               maxBridgeTokens=2 * 780, columns=dict(mirror='mirrorPosts', relays='bridgeRelays', bridgeTok='inTokPerCallBridge'),
               onFail='stop and ask'),
    costRule=dict(rule='report and wait (P20-8): the estimate before the pilot, the projection after it; the user approves each',
                  ceiling=35.0, projection="the pilot's cost per call by arm x c1Grid's calls (1,920 per liar board, 2,560 per clean)",
                  estimate=c1Range(), perCall=PER_CALL4, allowance=ALLOW20),
    cap=250.0, capNote='per calendar month (the user, 2026-11-09)',
    partB=dict(models=list(PARTB_MODELS), suite='squadEvalSuite: 32 agents in squads of 4, plain board, seeds 900-909, clean and 8 liars',
               reused='reusedPartB: |n32 and |n32liar of eval_p17 and eval_p17s0/warm.csv (180 rows)',
               weights='the final r17mix adapters as published (P20-10)'),
    check='bridgeCheckCells (P20-11): warm, N 16, world19e(S) + couple 0.25 + bridges 1, S 8 and 4, 4 guards, 4 liars, 900-904',
    dirs=dict(pilot='results/multiagent/colony/c1pilot', main='results/multiagent/colony/c1', partB='results/multiagent/colony/c1b'),
    nboot=2000, exploratory=True)
C1_PLAN = dict(   # frozen after the pilot and B3's forecast (T9); nothing in the hypotheses or the gate changes
    C1_PILOT_PLAN, frozen=False, pilotHash=None, verdict=None, projection=None,
    forecast=dict(file='specs/multiagent/phase20/c1_predictions.json', commit=None, sha256=None,
                  effects='b3_effects.C1_EFFECTS (B3, Phase 22): C1-1 to C1-8', lines=None,
                  how="B3's fitted simulator refit on the pilot, RuleBridge for bridges (P20-13); no fallback"))
```

## Cost rule and pilot check

```python
#| export
def costRule20(projected, plan=None):
    "P20-8, report and wait: the projection beside the expected ceiling; the verdict is always 'ask' (never 'run' on its own)."
    p = (plan or C1_PILOT_PLAN)['costRule']
    return dict(verdict='ask', projected=float(projected), ceiling=p['ceiling'], overCeiling=bool(projected > p['ceiling']),
                note=p['rule'])


def pilotCheck20(rows, plan=None):
    """FR-C2 on the pilot's rows (levels '<arm>|independent'): invalid replies under 5%, mirror posts above 0, relays above 0
    on each bridge board, a bridge's input tokens per call at most 2 x 780; then the projection of c1Grid at the pilot's cost
    per call by arm and costRule20 (always 'ask'). A missing column fails its check. -> dict(ok, reasons, values, ...)."""
    plan = plan or C1_PILOT_PLAN
    P, col = plan['pilot'], plan['pilot']['columns']
    f = lambda r, k: float(r[k]) if r.get(k) not in (None, '') else float('nan')
    by = {arm20(r): r for r in rows}
    calls = sum(f(r, 'calls') for r in rows) or 1.0
    bridged = [a for a in by if ARMS20[a].get('bridges')]
    vals = dict(invalid=sum(f(r, 'invalid') for r in rows) / calls, mirrorPosts=sum(f(r, col['mirror']) for r in rows),
                relays={a: f(by[a], col['relays']) for a in bridged}, bridgeTokens={a: f(by[a], col['bridgeTok']) for a in bridged})
    reasons = [m for m, bad in [
        (f"missing pilot boards {sorted(set(P['cells']) - set(by))}", set(P['cells']) - set(by)),
        (f"invalid {vals['invalid']:.3f} >= {P['maxInvalid']}", not vals['invalid'] < P['maxInvalid']),
        (f"no mirror posts ({vals['mirrorPosts']})", not vals['mirrorPosts'] > 0)] +
        [(f'no relays on {a} ({v})', not v > 0) for a, v in vals['relays'].items()] +
        [(f"{a}: bridge input tokens per call {v:.0f} > {P['maxBridgeTokens']}", not v <= P['maxBridgeTokens'])
         for a, v in vals['bridgeTokens'].items()] if bad]
    perCall = {a: f(r, 'cost') / f(r, 'calls') for a, r in by.items() if f(r, 'calls')}
    est = c1Estimate(c1Grid(), perCall) if perCall else dict(cost=float('nan'))
    return dict(ok=not reasons, reasons=reasons, values=vals, perCall=perCall, estimate=est, **costRule20(est['cost'], plan))
```

## Tests

```python
# test_c1_grid (FR-C2, FR-C3, FR-C4, AC-8, P20-8, P20-9), the plans' half: C1_PILOT_PLAN and C1_PLAN unfrozen and JSON-hashable,
# with cap 250 and its capNote and the $35 ceiling; C1_PLAN is the pilot plan plus the forecast, the verdict and the projection
# (placeholders until T9); pilotCheck20 and costRule20 on synthetic pilot rows: always 'ask'
from explRL.coop_grid import coopHash

for P in (C1_PILOT_PLAN, C1_PLAN):
    assert P['frozen'] is False and coopHash(P) and P['cap'] == 250.0 and P['capNote'] == 'per calendar month (the user, 2026-11-09)'
    assert P['costRule']['ceiling'] == 35.0 and P['plan'] == 'specs/multiagent/phase20/' and P['arms'] is ARMS20
assert {k: v for k, v in C1_PLAN.items() if k not in ('frozen', 'pilotHash', 'verdict', 'projection', 'forecast')} == \
    {k: v for k, v in C1_PILOT_PLAN.items() if k != 'frozen'} and C1_PLAN['forecast']['sha256'] is None
assert set(C1_PILOT_PLAN['hypotheses']) == {f'H20_{i}' for i in range(1, 9)} and 'forecast' not in C1_PILOT_PLAN
# pilotCheck20 and costRule20 on synthetic pilot rows: always 'ask'
row = lambda a, **k: {'level': f'{a}|independent', 'calls': 1920, 'invalid': 10, 'cost': 0.6, 'mirrorPosts': 200,
                      'bridgeRelays': 0 if a == 'coupled' else 50, 'inTokPerCallBridge': '' if a == 'coupled' else 900, **k}
ok = pilotCheck20([row(a) for a in PILOT20])
assert ok['ok'] and ok['verdict'] == 'ask' and ok['reasons'] == [] and ok['ceiling'] == 35.0
assert abs(ok['projected'] - (23 * 1920 + 15 * 2560) * 0.6 / 1920) < 1e-9 and not ok['overCeiling']
for k, v, a in (('invalid', 400, 'bridge1'), ('mirrorPosts', 0, None), ('bridgeRelays', 0, 'bridgeGuard'),
                ('inTokPerCallBridge', 1600, 'bridge1')):
    bad = pilotCheck20([row(x, **({k: v} if a in (None, x) else {})) for x in PILOT20])
    assert not bad['ok'] and bad['verdict'] == 'ask' and bad['reasons'], (k, bad)
bad = pilotCheck20([row(a) for a in PILOT20[:2]])
assert not bad['ok'] and 'missing' in bad['reasons'][0]
big = pilotCheck20([row(a, cost=2.0) for a in PILOT20])
assert big['ok'] and big['verdict'] == 'ask' and big['overCeiling']
assert {costRule20(x)['verdict'] for x in (0.0, 30.0, 35.0, 1e6)} == {'ask'}
```
